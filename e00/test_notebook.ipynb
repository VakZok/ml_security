```python
%load_ext autoreload
%autoreload 2
%env https_proxy=http://proxy:80
%env http_proxy=http://proxy:80


import numpy as np
import matplotlib.pyplot as plt
import torch
from tqdm import tqdm
import os


# Create tensor, move to CUDA, and back to numpy
cpu_tensor = torch.tensor([1.0, 2.0, 3.0])
cuda_tensor = cpu_tensor.to('cuda')
back_to_numpy = cuda_tensor.cpu().numpy()
print(f"Torch-CUDA-NumPy interplay working: {np.allclose(back_to_numpy, [1.0, 2.0, 3.0])}")

# Test tqdm
for _ in tqdm(range(10), desc="Testing tqdm"):
    pass

# Test matplotlib with random data from NumPy and PyTorch
np.random.seed(42)
torch.manual_seed(42)

np_samples = np.random.randn(10000)
torch_samples = torch.randn(10000).numpy()

ax = plt.subplot(111)
ax.hist(np_samples, bins=50, alpha=0.7, label='NumPy', density=True)
ax.hist(torch_samples, bins=50, alpha=0.7, label='PyTorch', color='orange', density=True)
ax.set_title('NumPy vs PyTorch Random Distribution')
ax.set_xlabel('Value')
ax.set_ylabel('Density')
ax.legend()
plt.tight_layout()
plt.show()

print("Matplotlib working: True")

if not os.path.exists('environment.yml'):
    raise FileNotFoundError("environment.yml file not found. Are you executing the notebook in the correct working directory? The working directory should be the one containing the environment.yml and this notebook.")
else:
    print("environment.yml file found: True, running in correct working directory.")
```
